```python
import gensim.downloader as api
#from gensim.models import FastText
#from gensim.test.utils import common_texts
import numpy as np
from tabulate import tabulate
import matplotlib.pyplot as plt
#import sent2vec
```

**Loading a word embedding (WE) model**

```python
word_vectors = api.load("glove-twitter-200")
```

**Calculating word-to-word similarities**

```python
#helper functions
def cosine_similarity(source_vec, target_matrix):
    dot_products = np.dot(target_matrix, source_vec)
    source_norm = np.linalg.norm(source_vec)
    target_norm = np.linalg.norm(target_matrix, axis=1)
    return dot_products / (source_norm * target_norm)

def dot_product_similarity(source_vec, target_matrix):
    return np.dot(target_matrix, source_vec)

def nearest_neighbors(source_word, target_words, k, metric="cosine"):
    source_vec = word_vectors[source_word]
    target_vecs = np.array([word_vectors[word] for word in target_words])
    
    if metric == "dot":
        similarities = dot_product_similarity(source_vec, target_vecs)
    elif metric == "cosine":
        similarities = cosine_similarity(source_vec, target_vecs)

    topk_idx = np.argsort(similarities)[::-1][:k]
    topk_words = [target_words[i] for i in topk_idx]
    topk_scores = [similarities[i] for i in topk_idx]
    
    return list(zip(topk_words, topk_scores))
```

**Calculating nearest neighbors**

```python
#source words
source_words = ["giraffe", "anthropology", "remarkable", "delicious", "pencil"]

#target words
target_words_all = {
    "giraffe": ["elephant", "lion", "cheetah", "zebra", "bicycle"],
    "anthropology": ["biology", "psychology", "engineering", "linguistics", "physics"],
    "remarkable": ["ambiguous", "disgraceful", "grumpy", "amazing", "ambitious"],
    "delicious": ["animated", "yummy", "moldy", "suitable", "flavorful"],
    "pencil": ["pen", "car", "desk", "sharpener", "paper"]
}

k = 10
results = []
for source_word in source_words:
    target_words = target_words_all[source_word]
    dot_nearest = nearest_neighbors(source_word, target_words, k, metric="dot")
    cosine_nearest = nearest_neighbors(source_word, target_words, k, metric="cosine")
    results.append({"source": source_word, "cosine similarity (k=10)": cosine_nearest, "dot similarity (k=10)": dot_nearest})
```

```python
output = []
for result in results:
    source = result["source"]
    cos_sim = ", ".join([f"{word} ({score:.3f})" for word, score in result["cosine similarity (k=10)"]])
    dot_sim = ", ".join([f"{word} ({score:.3f})" for word, score in result["dot similarity (k=10)"]])
    output.append([source, cos_sim, dot_sim])

print(tabulate(output, headers=["Source Word",
                                "Cosine Similarity (k=10)",
                                "Dot Product Similarity (k=10)"], tablefmt="grid"))
```

**Reporting and observations**

The results of both metrics capture the correct ranking of words, in Cosine Similarity the results are normalized, in Dot Product Similarity large values indicate high similarity.
The differences of rank in the results are for the words "anthropology", "pencil", and "remarkable" as expected, Dot Product Similarity is affected by the vector magnitudes.

Comparing the results of Cosine Similarity(CS) and Dot Product Similarity (DPS), I would prefer CS as it is computationally faster and works efficiently in high dimensional vector spaces. One major disadvantage is that is does not take into account the magnitute of the vectors in the sense that it is not sensitive to the length of documents.

In particular, given the document $d$, consisting of words $\left[ v_1, v_2, ..., v_{|d|} \right]$, the document representation $\mathbf{e}_d$ is defined as:

$\mathbf{e}_d = \frac{1}{|d|}\sum_{i=1}^{|d|}{\mathbf{e}_{v_i}}$

where $\mathbf{e}_{v}$ is the vector of the word $v$, and $|d|$ is the length of the document.
